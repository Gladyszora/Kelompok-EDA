Tugas Analisis 5: Setelah drop_duplicates(), berapa jumlah baris dataset sekarang dibanding  sebelumnya? Mengapa penting memastikan tipe data (dtypes) sudah benar sebelum data  dianalisis lebih lanjut?

jawab: jumlah baris dataset sebelum drop_duplicates() adalah 69 baris. setelah dilakukan drop_duplicates(), jumlah baris menjadi 65 baris karena 4 baris data duplikat telah dihapus. memastikan type dat sudah benar bertujuan untuk mencegah eror operasi matematika

```python
import pandas as pd

# 1. Baca dataset
df = pd.read_csv("dataset_penjualan_kantin.csv")

# 2. Hapus spasi nama kolom & hapus duplikat
df.columns = df.columns.str.strip()
df = df.drop_duplicates()

# 3. Hapus semua teks/huruf, ambil angkanya saja
df['jumlah_terjual'] = df['jumlah_terjual'].astype(str).str.replace(r'\D', '', regex=True)
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(r'\D', '', regex=True)

# 4. Ubah ke numerik, isi data kosong dengan 0, lalu ke integer
df['jumlah_terjual'] = pd.to_numeric(df['jumlah_terjual'], errors='coerce').fillna(0).astype(int)
df['harga_satuan'] = pd.to_numeric(df['harga_satuan'], errors='coerce').fillna(0).astype(int)

# 5. Hapus baris jika nama_produk kosong
df = df.dropna(subset=['nama_produk'])

# 6. Analisis data
laris = df[df['jumlah_terjual'] > 20]
urut = df.sort_values(by='jumlah_terjual', ascending=False)
df['total_pendapatan'] = df['harga_satuan'] * df['jumlah_terjual']
ringkasan = df.groupby('nama_produk')['total_pendapatan'].sum()

print(ringkasan)
```
